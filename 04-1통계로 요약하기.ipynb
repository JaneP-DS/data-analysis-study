####분위수 구하기

######ex) 이분위수는 전체 데이터를 두 구간으로 나누기 때문에 중앙값에 해당

```python
#1,2,3,4,5 다섯 개의 숫자가 있을 때 90% 위치에 있는 값은?
pd.Series([1,2,3,4,5]).quantile(0.9) #시리즈 객체를 정의한 후 분위수를 구함

#이렇게 두 지점 사이에 놓인 특정 위치의 값을 구하는 방법을 '보간'이라고 부름
```

```python
'''
설명
<quantile() 메서드>
- interpolation 매개변수에서 중간 값을 계산하는 방법을 결정함
* 만약 interpolation 매개변수를 따로 지정하지 않으면 기본값으로 계산됨 (기본값:linear/양쪽 분위수에 비례하여 결정됨)

<interpolation 매개변수의 또 다른 보간 방식>
pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='midpoint')
 ㄴ여기서 midpoint는 분위수에 상관없이 무조건 두 수 사의 중앙값을 사용함

 pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='nearest')
 ㄴ두 수 중에서 가까운 값을 선택함

 ** 이거 말고도 두 수 중 작은 값을 선택하는 'lower', 큰 값을 선택하는 'higher'가 있음 **
```

####평균 구하기(mean()함수, average()함수)

```python
#mean() 함수
import numpy as np
np.mean(ns_book7['대출건수'])     # 판다스와 동일
```

```python
#average() 함수
# - 기본적으로 mean()함수와 동일하게 평균을 계산하는데, weights 매개변수에 가중치를 제공하면 가중 평균을 계산함
#가중평균? : 평균을 구할 때 각 값의 중요도에 따라 가중치를 부여하여 계산하는 평균
# - 평균할 값에 각각의 가중치를 곱하고 전체 가중치로 나누어 계산함


#average() 함수로 평균 구하기(weight기준 : 도서권수의 역수로 지정)
np.average(ns_book7['대출건수'], weights=1/ns_book7['도서권수'])
```

####1/ns_book7['도서권수'] 가 어떻게 계산되는걸까?
-'브로드캐스팅' = 판다스 데이터프레임과 시리즈 객체에 스칼라 값으로 산술 연산을 수행하면 모든 원소에 연산이 동일하게 적용되는 과정을 통해 계산됨.

```python
#대출건수 열을 도서권수 열로 나누어 도서권수가 1일 때 대출건수를 얻기
np.mean(ns_book7['대출건수']/ns_book7['도서권수'])

#결과 : 모든 도서권수가 1이기 때문에 도서권수에 따라 대출건수에 미치는 영향을 덜 받아 기존 평균값 보다 낮아지게 됨
```

```python
#도서에 상관없이 한 권 당 대출건수 평균 구하기
ns_book7['대출건수'].sum()/ns_book7['도서권수'].sum()
```

####분산 구하기

```python
#var() 함수 제공
np.var(ns_book7['대출건수'])        #판다스와 다름(-> n-1 자유도 때문)
```

####**n-1 자유도**
-표본집단은 평균과 n-1개의 샘플 데이터를 알고 있다면 마지막 한 개의 샘플 값은 자동으로 알 수 있음. 이러한 경우를 자유도 n-1 이라고 함. 따라서 분산을 계산할 때 분모를 n이 아니라 n-1로 나눔

**이렇게 기억하자**

####판다스
-판다스 var()메서드의 ddof 매개변수 기본값은 1이므로 n-1로 나눗셈
####넘파이
-넘파이 var()메서드의 ddof 매개변수 기본값은 0이므로 n으로 나눗셈



```python
#판다스
ns_book7['대출건수'].var(ddof=0)
```

```python
#넘파이
np.var(ns_book7['대출건수'], ddof=1)
```

####최빈값 구하기

```python
#넘파이는 직접적으로 최빈값을 계산하는 함수를 제공하지 않음
#unique() 함수를 사용해 최빈값을 찾을 수 있음

#unique() 함수 : 배열에서 고유한 값을 찾아줌
#return_counts 매개변수 : 기본값 False에서 True로 바꿔 주면 고유한 값의 등장 횟수도 반환함

#먼저 unique() 함수를 호출해 고유한 값(values)와 등장 횟수(counts) 배열을 얻음
values, counts = np.unique(ns_book7['도서명'], return_counts=True)

#argmax() 함수 : 이 함수를 사용해 counts 배열에서 가장 큰 값의 인덱스 찾기
max_idx = np.argmax(counts)

#values 배열에서 해당 인덱스의 값을 출력함
values[max_idx]
```
