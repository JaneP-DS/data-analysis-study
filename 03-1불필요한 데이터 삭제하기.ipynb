##-원소별 비교
##### 판다스 배열 성격의 객체는 어떤 값과 비교할 때 자동으로 배열에 있는 모든 원소와 하나씩 비교해줌.

##-데이터프레임 중간에 있는 열 빼고 나머지를 선택하고 싶을 때
#####불리언 배열이나 drop() 메서드 사용하기

```python
#데이터프레임 중간에 있는 열을 삭제하고 싶을 때, 불리언 배열
ns_df.columns != '부가기호'
#반환된 결과는 '넘파이 배열'임
```

```python
#얻은 불리언 배열 값을 selected_columns 변수에 저장하고 판다스 데이터프레임의 loc 메서드에 전달하면 True인 열의 행만 선택가능
selected_columns = ns_df.columns != '부가기호'
ns_book = ns_df.loc[:, selected_columns]
ns_book.head()
```

```python
#'부가기호'와 'Unnamed: 13' 열 함께 제거(리스트 형식 사용)
ns_book = ns_df.drop(['부가기호','Unnamed: 13'], axis =1)
ns_book.head()
```

```python
#inplace() = True -> 현재 선택한 데이터프레임을 바로 수정할 수 있음.
#'주제분류번호'삭제하기
ns_book.drop('주제분류번호',axis=1,inplace=True)
ns_book.head()
```

```python
#dropna() 메서드 -> 기본적으로 NaN이 하나 이상 포함된 행이나 열을 삭제함
#how 매개변수를 'all'로 지정하면 모든값이 NaN인 열이 삭제됨
ns_book=ns_df.dropna(axis=1)
ns_book.head()
```

```python
#[]연산자와 슬라이싱
#[]연산자에 슬라이싱이나 불리언 배열을 전달하면 행을 선택함
ns_book2 = ns_book[2:] # 0,1을 제외한 행 출력
ns_book2.head()
```

#####중복된 행 찾기
#####-duplicated() 메서드 = 데이터프레임에 있는 모든 열을 기준으로 중복된 행을 찾음

```python
#duplicated() 메서드 -> 처음을 제외한 나머지 중복된 행을 True로 표시, 그 외 중복되지 않은 나머지 행들은 False로 표시
sum(ns_book.duplicated()) #sum() 함수를 함께 사용하면 True를 1로 인식하기 때문에 중복된 행의 개수를 셀 수 있음
```

```python
#일부 열을 기준으로 중복된 행을 찾으려면 subset 매개변수에 기준열을 나열함
sum(ns_book.duplicated(subset=['도서명','저자','ISBN']))
```

```python
#어떤 데이터가 중복되었는지 확인하기 = keep 매개변수(False로 지정하여 중복된 모든 행을 True로 표시한 불리언 배열을 반환)
dup_rows = ns_book.duplicated(subset=['도서명','저자','ISBN'], keep=False)
ns_book3 = ns_book[dup_rows]
ns_book3.head()
```

##### 그룹별로 모으기
#####-groupby() 메서드

```python
#groupby() 메서드의 by 매개변수에는 행을 합칠 때 기준이 되는 열을 지정함
count_df = ns_book[['도서명','저자','ISBN','권','대출건수']]

#groupby() 메서드는 기본적으로 by 매개변수에 지정된 열에 NaN이 포함되어 있으면 해당 해을 삭제함.
#이를 막기 위해 dropna 매개변수를 False로 지정함
group_df = count_df.groupby(by=['도서명','저자','ISBN','권'], dropna=False)

#sum() 메서드 = 대출건수를 하나로 합쳐야하므로 sum() 메서드를 활용
loan_count = group_df.sum()

loan_count = count_df.groupby(by=['도서명', '저자', 'ISBN','권'], dropna=False).sum()
#head() 메서드로 데이터프레임을 출력하면, 기준인 인덱스 열이 굵은 글시로 표시됨
loan_count.head()
```

#####원본 데이터 업데이트하기


1.   duplicated() 메서드로 중복된 행을 True로 표시한 불리언 배열을 만듦.
2.   1번에서 구한 불리언 배열을 반전시켜서 중복되지 않은 고유한 행을 True로 표시

-반전시킬 때는 판다스의 '~'연산자를 사용함

3.   2번에서 구한 불리언 배열을 사용해 원본 배열에서 고유한 행만 선택

-copy() 메서드로 ns_book3 데이터프레임을 만듦



```python
dup_rows=ns_book.duplicated(subset=['도서명','저자','ISBN','권']) # 1번

unique_rows = ~dup_rows # 2번
ns_book3 = ns_book[unique_rows].copy() # 3번
```

```python
# ns_book3에 중복된 행이 없는지 duplicated() 메서드로 확인하기
sum(ns_book3.duplicated(subset=['도서명','저자','ISBN','권']))
```

```python
#원본 데이터프레임 인덱스 설정하기(ns_book3의 인덱스를 loan_count 데이터프레임의 인덱스와 동일하게 만듦)
#set_index() -> 지정한 열을 인덱스로 설정할 때 사용
#inplace 매개변수를 True로 지정해 새로운 df을 반환하지 않고 ns_book3 df을 수정
ns_book3.set_index(['도서명','저자','ISBN','권'], inplace=True)
ns_book3.head()
```

```python
#업데이트하기 = update() 메서드
ns_book3.update(loan_count)
ns_book3.head()
```

```python
#업데이트가 되었다면 인덱스 열을 해제함 = reset_index() 메서드로 df 인덱스를 재설정가능
ns_book4 = ns_book3.reset_index()
ns_book4.head()
```

```python
#원본 데이터인 ns_book에서 대출건수가 100회 이상인 책의 개수 세기
sum(ns_book['대출건수']>100)
```

```python
#ns_book4에서 대출건수가 100회 이상인 책의 개수 세기
#늘어난 이유 : 중복된 도서의 대출건수를 합쳤기 때문
sum(ns_book4['대출건수']>100)
```

```python
#원래 열 이름대로 맞추기 = []연산자에 원래 df의 열 값을 전달
ns_book4 = ns_book4[ns_book.columns]
ns_book4.head()
```

```python
#일괄 처리 함수, data_cleaning() 함수
def data_cleaning(filename):
  """
  남산 도서관 장서 CSV 데이터 전처리 함수
  :param filename: CSV 파일 이름
  """

  #파일을 데이터프레임으로 읽습니다.
  ns_df = pd.read_csv(filename, low_memory=False)
  #NaN인 열을 삭제합니다.
  ns_book = ns_df.dropna(axis=1, how='all')
  #대출건수를 합치기 위해 필요한 행만 추출하여 count_df 데이터프레임을 만듭니다.
  count_df = ns_book[['도서명','저자','ISBN','권','대출건수']]
  #도서명, 저자, ISBN, 권을 기준으로 대출건수 groupby합니다.
  loan_count = count_df.groupby(by=['도서명','저자','ISBN','권'], dropna=False).sum()

  #원본 데이터프레임에서 중복된 행을 제외하고 고유한 행만 추출하여 복사합니다.
  dup_rows = ns_book.duplicated(subset=['도서명','저자','ISBN','권'])
  unique_rows = ~dup_rows
  ns_book3 = ns_book[unique_rows].copy()
  #도서명, 저자, ISBN, 권을 인덱스로 설정합니다.
  ns_book3.set_index(['도서명','저자','ISBN','권'], inplace=True)
  #loan_count에 저장된 누적 대출건수를 업데이트합니다.
  ns_book3.update(loan_count)
  #인덱스를 재설정합니다.
  ns_book4=ns_book3.reset_index()
  #원본 데이터프레임의 열 순서로 변경합니다.
  ns_book4 = ns_book4[ns_book.columns]
  return ns_book4
```

```python
#함수가 제대로 동작하는지 확인
new_ns_book4 = data_cleaning('ns_202604.csv')
#다른 데이터프레임을 비교할 때, equals() 메서드 사용
ns_book4.equals(new_ns_book4)
```
