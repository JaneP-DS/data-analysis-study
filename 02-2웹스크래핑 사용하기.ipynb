```python
#태그 안의 텍스트 가져오기, get_text()메서드
for tr in prd_tr_list: #for문으로 prd_tr_list를 순회하면서 <th> 태그 안의 텍스트가 '쪽수, 무게, 크기'인지 검사함
  if tr.find('th').get_text() == '쪽수, 무게, 크기': #get_text() 메서드(태그 안에 텍스트를 반환) = <td> 안에 있는 텍스트를 가져옴
    page_td = tr.find('td').get_text()
    break
```

##'전체 도서의 쪽수 구하기' - 총정리


1.   온라인 서점의 검색 결과 페이지 URL을 만듦
2.   requests.get()함수로 검색 결과 페이지의 HTML을 가져옴
3.   뷰티플수프로 HTML을 파싱함
4.   뷰티플수프의 find() 메서드로 a 태그를 찾아 상세 페이지 URL을 추출함
5.   requests.get() 함수로 다시 도서 상세 페이지의 HTML을 가져옴
6.   뷰티플수프로 HTML을 파싱함
7.   뷰티플수프의 find() 메서드로 '품목정보' div 태그를 찾음
8.   뷰티플수프의 find_all() 메서드로 '쪽수'가 들어있는 tr 태그를 찾음
9.   앞에서 찾은 테이블의 행에서 get_text() 메서드로 td 태그에 들어있는 '쪽수'를 가져옴

-> 이 작업을 구현한 get_page_cnt() 함수는 ISBN 정수 값을 받아 쪽수를 반환함



```python
def get_page_cnt(isbn):
  #Yes24 도서 검색 페이지 URL
  url = 'https://www.yes24.com/Product/Search?domain=ALL&query={}'

  #URL에 ISBN을 넣어 HTML을 가져옴
  r = requests.get(url.format(isbn))

  soup = BeautifulSoup(r.text, 'html.parser') #HTML 파싱

  # 검색 결과에서 해당 도서를 선택함
  prd_info = soup.find('a', attrs={'class':'gd_name'})
  if prd_info == None:
    return ''

  #도서 상세 페이지를 가져옴
  url = 'http://www.yes24.com'+prd_info['href']
  r = requests.get(url)
  soup = BeautifulSoup(r.text, 'html.parser')

  #품목정보 <div>를 선택
  prd_detail = soup.find('div', attrs={'id':'infoset_specific'})

  #테이블에 있는 <tr> 태그를 가져옴
  prd_tr_list = prd_detail.find_all('tr')

  #쪽수가 들어 있는 <th>를 찾아 <td>에 담긴 값을 반환함
  for tr in prd_tr_list:
    if tr.find('th').get_text() == '쪽수, 무게, 크기':
      return tr.find('td').get_text().split()[0]
  return ''
```

```python
get_page_cnt(9791189467302)
```

```python
#데이터프레임 행 혹은 열에 함수 적용하기, apply() 메서드
top10_books = books.head(10)
```

```python
def get_page_cnt2(row): #데이터프레임의 한 행을 받아 get_page_cnt() 함수를 사용해 쪽수 구하는 함수를 만듦
  isbn = row['isbn13']
  return get_page_cnt(isbn)
```

```python
page_count = top10_books.apply(get_page_cnt2, axis = 1) # apply()메서드의 첫 번째 매개변수=실행할 함수 & axis = 1 행에 함수를 적용
print(page_count) # get_page_cnt2()함수의 결괏값은 page_count 변수에 판다스 시리즈 객체로 저장됨
```

####람다 함수(lambda)
-> 람다 함수는 함수 이름 없이 한 줄로 쓰는 함수임

```python
#람다 함수 활용
page_count = top10_books.apply(lambda row: get_page_cnt(row['isbn13']), axis=1)
```

```python
print(page_count)
```

```python
#page_count 시리즈 객체에 이름 지정해주기(시리즈 객체에 name 속성을 사용하면 이름을 간단하게 지정할 수 있음)
page_count.name = 'page_count'
print(page_count)
```

```python
#데이터프레임과 시리즈 합치기, merge()함수
#두 객체의 인덱스를 기준으로 합칠 경우 left_index와 right_index 매개변수를 True로 지정함
top10_with_page_count = pd.merge(top10_books, page_count, left_index=True, right_index=True)
top10_with_page_count
```
