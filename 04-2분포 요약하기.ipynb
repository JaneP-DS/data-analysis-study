####히스토그램 그리기
- 히스토그램은 수치형 특성의 값을 일정한 구간으로 나누어 구간(=계급) 안에 포함된 데이터 개수를 막대 그래프로 그린 것

-도수 : 구간 안에 속한 데이터 개수

-도수분포표 : 구간과 도수를 표로 요약한 것

```python
#hist() 함수 : 히스토그램을 그리는 함수
#- hist() 함수는 1차원 데이터를 입력받아 히스토그램을 그리며, 기본적으로 데이터를 10개의 구간으로 나눔
plt.hist([0,3,5,6,7,7,9,13], bins=5)     #bins 매개변수 : 구간을 지정
plt.show()
```

```python
#넘파이에서 제공하는 histogram_bin_edges() 함수를 통해 구간의 경곗값 확인하기
import numpy as np
np.histogram_bin_edges([0,3,5,6,7,7,9,13], bins=5)

#결과 해석
#- 0~2.6 = 첫 번째 구간 / 2.6~5.2 = 두 번째 구간 ... 이런식으로 해석됨
```

```python
#넘파이의 randn() 함수 : 표준정규분포를 따르는 랜덤한 실수를 생성할 수 있음
#seed() 함수를 사용하면 '유사난수(=가짜 난수)'를 생성할 수 있음 / randn() 함수 전에 seed() 함수를 사용하면 항상 같은 난수를 추출할 수 있음
np.random.seed(42)
random_samples = np.random.randn(1000)
```

####정규분포와 표준정규분포
-종 모양처럼 가운데가 볼록하고 평균을 중심으로 대칭인 분포를 '정규분포'라고함. 특히 평균이 0이고 표준편차가 1인 정규분포를 '표준정규분포'라고 함

```python
#1,000개의 실수로 이루어진 random_samples 배열이 표준정규분포를 따르고 있는지 확인하기
#평균과 표준편차 계산
print(np.mean(random_samples), np.std(random_samples))

#결과해석
#- 평균이 약 0.02, 표준편차가 약 0.98 이므로 표준정규분포를 따르고 있다고 볼 수 있음
```

```python
plt.hist(random_samples)
plt.show()
```

####구간 조정하기

####상자 수염 그림 그리기
- 상자 수염 그림은 최솟값, 세 개의 사분위수, 최댓값 이렇게 다섯 개의 숫자를 사용해 데이터를 요약하는 그래프를 그림

**순서**

1. 사분위수를 계산하여 25%와 75% 지점을 밑면과 윗면으로 하는 직사각형을 그림

2. 중간값(50%)에 해당하는 지점에 수평선을 그음

3. 사각형의 밑면과 윗면에서 사각형의 높이의 1.5배만큼 떨어진 거리 안에서 가장 멀리 있는 샘플까지 수직선을 그음

4. 이 수직선(수염) 밖에서 최솟값과 최댓값까지 데이터를 점으로 표시함. 이 값을 '이상치'라고 부름


-IQR : 제1사분위수와 제3사분위수 사이의 거리를 말함


**장점**

-상자 수염 그림은 여러 개의 특성을 시각적으로 비교하기 좋음. 특히 데이터가 어떤 방향으로 더 많이 늘어져 있는지 한눈에 파악가능

```python
#boxplot() 함수 : 상자 수염 그림 그리기
plt.boxplot(ns_book7[['대출건수','도서권수']])
plt.show()

#대출건수와 도서권수 데이터는 사분위수가 매우 작아서 직사각형 상자가 거의 보이지 않음. 중간 값은 붉은 수평선으로 표시됨
```

```python
#y축을 로그 스케일로 바꾸기
plt.boxplot(ns_book7[['대출건수','도서권수']])       #겉에 []는 DF에서 열을 선택하는 문법, 안쪽 []는 선택할 열 이름을 리스트로 묶은 것
plt.yscale('log')
plt.show()

#결과
#-도서권수 열의 사분위수는 모두 1이기 때문에 상자가 보이지 않는 것이 맞음
```

####상자 수염 그림 수평으로 그리기

```python
#boxplot() 함수의 vert() 매개변수를 기본값 True에서 False로 바꾸면 됨
plt.boxplot(ns_book7[['대출건수','도서권수']], vert=False)
plt.xscale('log')       #x와 y축이 바뀌므로 로그 스케일도 x축에 지정해야함
plt.show()
```

####수염 길이 조정하기

```python
#기본적으로 수염 길이는 IQR의 1.5배임
#boxplot() 함수의 whis 매개변수에서 이를 조정가능(기본값 : 1.5)
plt.boxplot(ns_book7[['대출건수','도서권수']], whis=10)
plt.yscale('log')
plt.show()

#결과해석
#-1번 상자는 길어졌지만 2번은 IQR이 0이기 때문에 전과 같음
```

```python
#whis() 매개변수는 백분율로도 지정가능함. 예를 들어 whis(10.90)이면 10%, 90% 백분위수에 해당하는 데이터까지 수염을 그림
plt.boxplot(ns_book7[['대출건수','도서권수']], whis=(0,100))     #100%로 하면 마지막 데이터까지 수염을 그릴 수 있음
plt.yscale('log')
plt.show()
```

```python
#plot 속성 : 판다스 데이터프레임 객체의 plot 속성은 여러 가지 그래프를 그릴 수 있는 메서드를 제공함
#scatter() 메서드 : 산점도 그리기
ns_book7.plot.scatter('도서권수','대출건수', alpha=0.1)      #x축과 y축에 해당하는 열 이름만 지정하면 됨
plt.show()
```

####히스토그램 그리기
