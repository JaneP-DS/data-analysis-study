```python
d={"name": "혼자 공부하는 파이썬"} #JSON 형식은 키와 값에 큰따옴표를 사용함
print(d['name']) #파이썬은 둘 다 사용 가능
```

```python
#json 패키지를 활용해 d를 JSON 형식에 맞는 문자열로 바꾸기
import json
```

```python
#PYTHON -> JSON, json.dumps()함수 사용
#웹 기반 API는 전송하려는 파이썬 객체를 JSON 문자열로 변환하여 전송함
d_str = json.dumps(d, ensure_ascii=False) #'ensure_ascii=False' = 한글이 포함되어있는 d를 원래 저장된 문자로 볼 수 있게 하기 위해서 사용함
print(d_str)
```

```python
print(type(d_str))
```

```python
#JSON -> PYHTON, json.loads()함수 사용
#JSON 문자열을 다시 파이썬 프로그램에 쓰려면 딕셔너리로 바꿔야함
d2 = json.loads(d_str)
print(d2['name'])
```

```python
print(type(d2))
```

```python
#XML 문자열 -> 파이썬 객체로 변환, fromstring() 함수
x_str = """
<book>
  <name>혼자 공부하는 파이썬</name>
  <author>박해선</author>
  <year>2022</year>
</book>
"""
```

```python
#fromstring()함수가 반환하는 객체는 ElementTree 모듈 아래 정의된 Element 클래스의 객체임
import xml.etree.ElementTree as et
book = et.fromstring(x_str)
```

```python
print(type(book))
```

```python
print(book.tag) #book 객체의 tag 속성을 출력하면 엘리먼트 이름을 확인할 수 있음
```

```python
#자식 엘리먼트 확인
book_childs = list(book) #list로 변환
print(book_childs)
```

```python
#book_childs 변수에 각 항목을 할당하고 text 속성으로 엘리먼트에 있는 텍스트를 출력(순서를 보장하진 않음)
name, author, year = book_childs
print(name.text)
print(author.text)
print(year.text)
```

```python
#findtext() 메서드 -> 해당하는 자식 엘리먼트를 탐색하여 자동으로 텍스트를 반환(자식 엘리먼트의 순서가 어떻든 상관없음)
name = book.findtext('name')
author = book.findtext('author')
year = book.findtext('year')
print(name)
print(author)
print(year)
```

```python
#자식 엘리먼트가 중복일 때는 부모 엘리먼트를 복수형으로 써주는걸 권장
x2_str = """
<books>
  <book>
    <name>혼자 공부하는 파이썬</name>
    <author>박해선</author>
    <year>2022</year>
  </book>

  <book>
    <name>혼자 공부하는 머신러닝</name>
    <author>박해선</author>
    <year>2020</year>
  </book>

</books>
"""
```

```python
import xml.etree.ElementTree as et

books = et.fromstring(x2_str)
print(books.tag)
```

```python
#여러 개의 자식 엘리먼트 확인하기, findall() 메서드와 for문
#동일한 이름을 가진 여러 개의 자식 엘리먼트를 찾을 때는 findall()메서드와 for문을 함께 사용하면 편함ㄴㄴ
for book in books.findall('book'):
  name = book.findtext('name')
  author = book.findtext('author')
  year = book.findtext('year')
  print(name)
  print(author)
  print(year)
  print()
```

```python
#리스트 내포 설명(for문을 대괄호로 감싸서 리스트로 만드는 방법)
#예시1
#기존 for문
squares = []
for x in range(1, 6):
    squares.append(x**2)

print(squares)  # [1, 4, 9, 16, 25]

#리스트 내포
squares = [x**2 for x in range(1, 6)]

print(squares)  # [1, 4, 9, 16, 25]

#------------------------------------------------------------------------------------------

#예시2
#기존 for문 & if문
evens = []
for x in range(1, 11):
    if x % 2 == 0:
        evens.append(x**2)

# 리스트 내포
evens = [x**2 for x in range(1, 11) if x % 2 == 0]

print(evens)  # [4, 16, 36, 64, 100]
```
