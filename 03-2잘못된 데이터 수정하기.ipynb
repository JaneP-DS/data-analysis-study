#### replace() 메서드에 정규 표현식 사용하기
###정규표현식(=정규식)이란?
- 문자열 패턴을 찾아서 대체하기 위한 규칙의 모음

```python
#숫자 찾기 : \d
#ex) 2026 -> \d\d\d\d
#그룹으로 묶기 \d\d(\d\d) -> \1,\2로 그룹을 나타냄
#regex = True -> 정규 표현식을 사용한다는 의미
#r = 원시 문자열을 만듦
ns_book4.replace({'발행년도':{r'\d\d(\d\d)':r'\1'}}, regex=True)[100:102]
```

```python
#정규 표현식이 반복될 때는 중괄호를 사용하여 개수를 지정할 수 있음
#\d{2} = \d\d (같은 표현)
ns_book4.replace({'발행년도':{r'\d{2}(\d{2})':r'\1'}},regex=True)[100:102]
```

```python
#문자 찾기 : 마침표(.)
ns_book4.replace({'저자':{r'(.*)\s\(지은이\)(.*)\s\(옮긴이\)': r'\1\2'}, '발행년도': {r'\d{2}(\d{2})': r'\1'}}, regex=True)[100:102]
```

```python
#뷰티플수프를 사용해 값 채우기
import requests
from bs4 import BeautifulSoup
```

```python
def get_book_title(isbn):
  url = 'https://www.yes24.com/product/search?domain=ALL&query=%25EC%259A%25B0%25EB%25A6%25AC%25EA%25B0%2580%2520%25EB%25B9%259B%25EC%259D%2598%2520%25EC%2586%258D%25EB%258F%2584%25EB%25A1%259C%2520%25EA%25B0%2588%2520%25EC%2588%2598%2520%25EC%2597%2586%25EB%258B%25A4%25EB%25A9%25B4'
  r=requests.get(url.format(isbn))
  soup = BeautifulSoup(r.text, 'html.parser')

  title = soup.find('a', attrs={'class':'gd_name'}).get_text()

  return title
```

```python
#저자, 출판사, 발행 연도를 추출하여 반환하는 함수 만들기
import re

def get_book_info(row):
  title = row['도서명']
  author = row['저자']
  pub = row['출판사']
  year = row['발행년도']

  url = 'https://www.yes24.com/product/search?domain=ALL&query=%25EC%259A%25B0%25EB%25A6%25AC%25EA%25B0%2580%2520%25EB%25B9%259B%25EC%259D%2598%2520%25EC%2586%258D%25EB%258F%2584%25EB%25A1%259C%2520%25EA%25B0%2588%2520%25EC%2588%2598%2520%25EC%2597%2586%25EB%258B%25A4%25EB%25A9%25B4'

  r=requests.get(url.format(row['ISBN']))
  soup = BeautifulSoup(r.text, 'html.parser')
  try:
    if pd.isna(title):
      title = soup.find('a', attrs={'class':'gd_name'}).get_text()

  except AttributeError:
    pass

  try:
    if pd.isna(author):
      authors = soup.find('span', attrs={'class':'info_auth'}).find_all('a')
      author_list = [auth.get_text() for auth in authors]
      author = ','.join(author_list)

  except AttributeError:
    pass

  try:
    if pd.isna(pub):
      pub = soup.find('span', attrs={'class':'info_pub'}).find('a').get_text()

  except AttributeError:
    pass

  try:
    if year == -1:
      year_str = soup.find('span', attrs={'class':'info_date'}).get_text()

      year = re.findall(r'\d{4}', year_str)[0]

  except AttributeError:
    pass

  return title, author, pub, year
```

```python
#누락된 값이 있었던 처음 두 개의 행에 get_book_info()함수 적용하기
#apply() 메서드는 기본적으로 반환된 값을 하나의 튜플로 만듦
#result_type 매개변수를 'expand'로 지정하여 반환된 값을 각기 다른 열로 만들 수 있음

updated_sample = ns_book5[na_rows].head(2).apply(get_book_info, axis=1, result_type = 'expand')
updated_sample
```

#### 일괄 처리 함수

```python
def data_fixing(ns_book4):


#잘못된 값을 수정하거나 NaN을 채우는 함수
#:param ns_book4: data_cleaning()함수에서 전처리된 데이터프레임

#도서권수와 대출건수를 int64로 바꿈
ns_book4 = ns_book4.astype({'도서권수':'int64','대출건수':'int64'})
#NaN인 세트 ISBN을 빈 문자열로 바꿈
set_isbn_na_rows = ns_book4['세트 ISBN'].isna()
ns_book4.loc[set_isbn_na_rows, '세트 ISBN']=''
#발행년도 열에서 연도 네 자리를 추출하여 대체함. 나머지 발행년도는 -1로 바꿈
ns_book5 = ns_book4.replace({'발행년도':r'.*(\d{4}).*'}, r'\1', regex=True)
unknown_year = ns_book5['발행년도'].str.contains(r'\D', na=True)
ns_book5.loc[unknown_year, '발행년도'] = '-1'
#발행년도를 int32로 바꿈
ns_book5 = ns_book5.astype({'발행년도':'int32'})
#4000년 이상인 경우 2333년을 뺌
dangun_yy_rows = ns_book5['발행년도'].gt(4000)
ns_book5.loc[dangun_yy_rows, '발행년도'] = ns_book5.loc[dangun_yy_rows, '발행년도']-2222

#여전히 4000년 이상인 경우 -1로 바꿈
dangun_year = ns_book5['발행년도'].gt(4000)
ns_book5.loc[dangun_year, '발행년도'] = -1

#0~1900년 사이의 발행년도는 -1로 바꿈
old_books = ns_book5['발행년도'].gt(0) & ns_book5['발행년도'].lt(1900)
ns_book5.loc[old_books, '발행년도']= -1

#도서명, 저자, 출판사가 NaN이거나 발행년도가 -1인 행을 찾음
na_rows = ns_book5['도서명'].isna() | ns_book5['저자'].isna()\
          |ns_book5['출판사'].isna() | ns_book5['발행년도'].eq(-1)

#Yes24 도서 상세 페이지에서 누락된 정보를 채움
updated_sample= ns_book5[na_rows].apply(get_book_info, axis =1, result_type = 'expand')
updated_sample.columns = ['도서명','저자','출판사','발행년도']
ns_book5.update(updated_sample)

#도서명, 저자, 출판사가 NaN이거나 발행년도가 -1인 행을 삭제함
ns_book6 = ns_book5.dropna(subset=['도서명','저자','출판사'])
ns_book6 = ns_book6[ns_book6['발행년도']!= -1]
```
